# Liquid Stacking Mean Reversion strategy

Backtest the strategy using the dedicated `LiquidStackingMeanReversion` pipeline

The strategy is being tested on several combination of markets and tokens (aka combination). Each combination is referenced as `M1_TOKEN1_M2_TOKEN2`, where :
 - `M1` is the reference market for modelling and inferences
 - `TOKEN1` is the token traded on `M1`
 - `M2` is the alternative market for modelling and inferences
 - `TOKEN2` is the token traded on `M2`

```python
import sys
sys.path.append("./src")
```

```python
import backtest_pipelines as bp
```

## 1. Create a new pipeline

Instantiate a new `LiquidStackingMeanReversion` object and load the fitted statistcal models

The statistical models are stored in a pickled Pandas DataFrame with one model per combination.

These are all FIGARCH models with AR means where the residuals are assumed to follow a fitted Student-t distribution. On the distribution of residuals, some combinations exhibit a different behavior on either one or both of the tails. For these, a Generalized Pareto Distribution has been  fitted on the concerned tail(s), allowing extreme values inference. 

During the statistical studies it was also evidenced that some combinations showed unstabilities on one-side of the residual distribution. As the inference would not be reliable, opening a new position on such side is effectively restricted (i.e. when the left-side is unstable it is prohibited to open a Long position on M1,PAIR1 combined with a Short position on M2,PAIR2 and vice-versa)

```python
strat = bp.LiquidStackingMeanReversion()
strat.import_models("./figarch_1_1_delta_fitted_viable_only.pickle")
```

## 2. Display and check the strategy parameters

Parameters can be modified using the corresponding setter (e.g to update `MIN_PROBABILITY_ENTRY_SIGNAL` to `0.4`, run `strat.set_min_probability_entry_signal(0.4)`)

```python
strat.show_params()
```

# 3. Run the pipeline and check the results

Tests the strategy on each combination using the corresponding statistical model. To make the full execution faster the pipeline can be run as multiprocessed which tests 1 combination per process.

Results are saved in subfolders of the path provided in the `OUTPUT_BASEPATH` parameter

```python
strat.run(num_workers=3)
```
